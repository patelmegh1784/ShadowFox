```python
import pandas as pd

# Load the Excel file and read Sheet2
file_path = "IPL sample data (1).xlsx"
df = pd.read_excel(file_path, sheet_name='Sheet2')

# Replace NaN with default values for processing
df.fillna('', inplace=True)

# Define weights for each fielding metric (adjust if needed)
weights = {
    'CP': 1,     # Clean Pick
    'GT': 2,     # Good Throw
    'C': 10,     # Catch
    'DC': -5,    # Dropped Catch
    'ST': 8,     # Stumping
    'RO': 7,     # Run Out
    'MRO': -3,   # Missed Run Out
    'DH': 4,     # Direct Hit
}

# Define function to calculate performance score for each player
def calculate_ps(player_df):
    cp = (player_df['Pick'] == 'clean pick').sum()
    gt = (player_df['Pick'] == 'good throw').sum()
    c = (player_df['Pick'] == 'catch').sum()
    dc = (player_df['Pick'] == 'drop catch').sum()
    st = (player_df['Throw'] == 'stumping').sum()
    ro = (player_df['Throw'] == 'run out').sum()
    mro = (player_df['Throw'] == 'missed run out').sum()
    dh = (player_df['Short Description'].str.contains('direct hit', case=False)).sum()
    rs = player_df['Runs'].sum()

    ps = (cp * weights['CP'] + gt * weights['GT'] + c * weights['C'] +
          dc * weights['DC'] + st * weights['ST'] + ro * weights['RO'] +
          mro * weights['MRO'] + dh * weights['DH'] + rs)
    return ps

# List of players to analyze (example: top 3 unique players from data)
players = df['Player Name'].unique()[:3]

# Calculate PS for each player
results = []
for player in players:
    player_df = df[df['Player Name'] == player]
    ps = calculate_ps(player_df)
    results.append({'Player': player, 'Performance Score': ps})

# Create DataFrame of results and display
results_df = pd.DataFrame(results)
print(results_df)
```
